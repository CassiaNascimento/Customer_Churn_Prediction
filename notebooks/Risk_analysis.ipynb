#### Extra analysis: Phone Service + Internet Service

Customers who subscribe to more than one service have a greater impact on the company's revenue and, as shown previously, exhibit a higher churn rate. Therefore, it is worthwhile to compare the characteristics of this subgroup with those of the overall customer population.

```python
df_reduced=df[(df["PhoneService"]=="Yes") & (df["InternetService"]!="No")]
df_reduced.head()
```

##### Comparing across each categorical feature

```python
categorical_features=[col for col in df.columns if df[col].dtype=='object']
```

```python
categorical_features
```

```python
#To enable direct comparisons of category proportions, we exclude the options listed below. For example, when comparing the proportions of customers using Fiber Optic and DSL internet services, 
#the overall customer population should include only customers with an internet service. Otherwise, the comparison would be distorted by customers without internet service. Under this approach, 
#for this particular example, the only difference between the two groups is that the overall customer population also includes customers who subscribe exclusively to internet service.

exclude = {
    "MultipleLines": "No phone service",
    "InternetService": "No",
    "OnlineSecurity": "No internet service",
    "OnlineBackup": "No internet service",
    "DeviceProtection": "No internet service",
    "TechSupport": "No internet service",
    "StreamingTV": "No internet service",
    "StreamingMovies": "No internet service",
}

for col in categorical_features:
    if col in exclude:
        value=exclude[col]
        df_trimmed=df[df[col]!=value]
        df_reduced_trimmed=df_reduced[df_reduced[col]!=value]
    else:
        df_trimmed=df
        df_reduced_trimmed=df_reduced

    comparison=pd.concat((df_trimmed[col].value_counts(normalize=True),df_reduced_trimmed[col].value_counts(normalize=True)),axis=1, keys=["All customers", "Subgroup"])*100
    display(comparison)
```

#### Summary: 
Customers subscribed to more than one service form a subgroup with 32.8% churn rate: the combination Internet Service + Phone Service comprises 4832 customers (68.7% of total customers).

Compared with the overall customer population, this subgroup presents the following characteristics:

- No significant differences observed in the proportion of male/female, number of customers with partners, customers with/without online security, online backup, device protection, tech support, streaming TV and streaming Movies.

- Higher proportion of senior citizens in this group: 20.4% compared with 16.2% in the overall population

- Lower proportion of customers with dependents in this group: 26% compared with 29.8% in the overall population

- Higher proportion of customers with multiple lines: 54.4% compared with 46.7% in the overall population

- Higher proportion of fiber optic internet: 64%(Fiber)/36%(DSL) compared with 56%(Fiber)/44%(DSL) in the overall population

- There is a higher proportion of month to month contracts and lower 1y, 2y, (m-m/1y/2y): 61.6%/20%/18.5% compared with 55%/21%/24% in the overall population

- Higher proportion of paperless billing: 69% compared with 59% in the overall population

- The proportion of Electronic check payments grew, while mailed check payments reduced, they are 41.7% (Electronic check) 14.7% (Mailed check) compared with 33.6% (Electronic check) 22.8% (Mailed check) in the overall population

- Higher churn rate: 32.8% compared with 26.6% in the overall population

- Similar tenure average(median) 33(30) months compared with 32.4(29) months in the overall population

- Higher monthly charges average(median) \\$81.76(\\$82.5) compared with \\$64.8(\\$70.35) in the overall population

- Higher total charges average(median) \\$2902.5(\\$2350.6) compared with \\$2283.3(\\$1397.5) in the overall population

These observations motivate a more detailed analysis of the churn rate as a function of the number of services subscribed to by each customer. For a deeper investigation, see the Risk Analysis by Number of Services notebook.
